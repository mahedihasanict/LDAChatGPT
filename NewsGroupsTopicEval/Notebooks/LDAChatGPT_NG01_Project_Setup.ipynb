```python
# ============================================================
# Imports
# ============================================================

import os
import json
import random

from pathlib import Path

import numpy as np
```

```python
# ============================================================
# Project Root
# ============================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/LDAChatGPT/NewsGroupsTopicEval"
)

print("Project root:")
print(PROJECT_ROOT)
```

```python
# ============================================================
# Create Project Directory Structure
# ============================================================

DIRECTORIES = {

    "notebooks":
        PROJECT_ROOT / "Notebooks",

    "data":
        PROJECT_ROOT / "data",

    "folds":
        PROJECT_ROOT / "folds",

    "models":
        PROJECT_ROOT / "models",

    "topic_vectors":
        PROJECT_ROOT / "topic_vectors",

    "intrinsic_metrics":
        PROJECT_ROOT / "intrinsic_metrics",

    "classification":
        PROJECT_ROOT / "classification",

    "statistics":
        PROJECT_ROOT / "statistics",

    "analysis":
        PROJECT_ROOT / "analysis",

    "figures":
        PROJECT_ROOT / "figures",

    "tables":
        PROJECT_ROOT / "tables",

    "logs":
        PROJECT_ROOT / "logs",

    "config":
        PROJECT_ROOT / "config",

    "dictionary":
        PROJECT_ROOT / "dictionary"
}


for directory in DIRECTORIES.values():

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


print("✓ Project directory structure created.")
```

```python
# ============================================================
# Global Experimental Configuration
# ============================================================

CONFIG = {

    # --------------------------------------------------------
    # Project
    # --------------------------------------------------------

    "project_name":
        "NewsGroupsTopicEval",

    "dataset_name":
        "20_Newsgroups",


    # --------------------------------------------------------
    # Dataset
    # --------------------------------------------------------

    "dataset_source":
        "scikit-learn",

    "dataset_api":
        "fetch_20newsgroups",

    "text_column":
        "Text",

    "label_column":
        "Label",

    "label_id_column":
        "LabelID",

    "document_id_column":
        "DocumentID",


    # --------------------------------------------------------
    # Reproducibility
    # --------------------------------------------------------

    "random_seed":
        42,


    # --------------------------------------------------------
    # Cross Validation
    # --------------------------------------------------------

    "n_folds":
        5,


    # --------------------------------------------------------
    # Dictionary Filtering
    # --------------------------------------------------------

    "remove_below":
        5,

    "remove_above":
        0.5,


    # --------------------------------------------------------
    # Topic Range
    # --------------------------------------------------------

    "topic_min":
        2,

    "topic_max":
        100,


    # --------------------------------------------------------
    # LDA Parameters
    # --------------------------------------------------------

    "lda_passes":
        20,

    "lda_iterations":
        400,

    "lda_chunksize":
        2000,

    "lda_alpha":
        "symmetric",

    "lda_eta":
        "symmetric",

    "lda_update_every":
        1,

    "lda_decay":
        0.5,

    "lda_offset":
        1.0,

    "lda_minimum_probability":
        0.0,

    "lda_per_word_topics":
        False,

    "lda_eval_every":
        None,

    "lda_dtype":
        "float32",


    # --------------------------------------------------------
    # Topic Representation
    # --------------------------------------------------------

    "top_words":
        20
}
```

```python
# ============================================================
# Save Configuration
# ============================================================

CONFIG_FILE = (
    PROJECT_ROOT
    / "config"
    / "config.json"
)


with open(
    CONFIG_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        CONFIG,
        file,
        indent=4
    )


print("✓ Configuration saved successfully.")

print(CONFIG_FILE)
```

```python
# ============================================================
# Set Random Seeds
# ============================================================

SEED = CONFIG["random_seed"]


random.seed(SEED)

np.random.seed(SEED)

os.environ["PYTHONHASHSEED"] = str(SEED)


print(
    f"✓ Random seed set to: {SEED}"
)
```
